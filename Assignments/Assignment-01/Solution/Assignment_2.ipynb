```python
import math
import csv
from collections import defaultdict
from heapq import heappush, heappop

COORDINATES_FILE = "Coordinates.csv"
DISTANCE_FILE = "distances.csv"



def heuristic(star1, star2):
    x1, y1, z1 = coordinates[star1]
    x2, y2, z2 = coordinates[star2]

    return math.sqrt(
        (x1 - x2) ** 2 +
        (y1 - y2) ** 2 +
        (z1 - z2) ** 2
    )



coordinates = {}

with open(COORDINATES_FILE, "r") as file:
    reader = csv.reader(file)
    next(reader)

    for star, x, y, z in reader:
        coordinates[star] = (
            int(x),
            int(y),
            int(z)
        )



adjacency_list = defaultdict(list)

with open(DISTANCE_FILE, "r") as file:
    reader = csv.reader(file)

    for source, destination, dist in reader:
        adjacency_list[source].append(
            (destination, int(dist))
        )



SOURCE_STAR = "Sun"
DESTINATION_STAR = "61 Virginis"

# DESTINATION_STAR = "Upsilon Andromedae"
# SOURCE_STAR = "TRAPPIST-1"
# DESTINATION_STAR = "55 Cancri"



def a_star(start, goal):

    priority_queue = []

    heappush(
        priority_queue,
        (heuristic(start, goal), 0, start)
    )

    parent = {}
    g_cost = {start: 0}

    visited = set()

    while priority_queue:

        f, current_g, current = heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:

            path = []

            while current in parent:
                path.append(current)
                current = parent[current]

            path.append(start)
            path.reverse()

            print("Path:")
            print(" -> ".join(path))

            print("\nTotal Distance:", current_g)

            return

        for neighbor, distance in adjacency_list[current]:

            new_g = current_g + distance

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                h = heuristic(neighbor, goal)

                f = new_g + h

                parent[neighbor] = current

                heappush(
                    priority_queue,
                    (f, new_g, neighbor)
                )

    print("Destination not reachable.")


a_star(SOURCE_STAR, DESTINATION_STAR)
```
