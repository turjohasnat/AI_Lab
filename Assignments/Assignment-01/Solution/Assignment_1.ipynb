```python
import math
import heapq


# State Class
class NodeState:
    def __init__(self, node_id, parent, g, f):
        self.node_id = node_id
        self.parent = parent
        self.g = g
        self.f = f

    def __lt__(self, other):
        return self.f < other.f


# Heuristic Function
def heuristic(node, goal, coords):
    x1, y1 = coords[node]
    x2, y2 = coords[goal]

    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)


# A* Search
def a_star_search(start, goal, coords, adjlist):

    minQ = []

    start_h = heuristic(start, goal, coords)

    start_state = NodeState(
        start,
        None,
        0,
        start_h
    )

    heapq.heappush(minQ, start_state)

    best_cost = {}

    while minQ:

        curr = heapq.heappop(minQ)

        # Skip if we already found a cheaper path
        if curr.node_id in best_cost and curr.g >= best_cost[curr.node_id]:
            continue

        best_cost[curr.node_id] = curr.g

        # Goal reached
        if curr.node_id == goal:

            path = []

            while curr is not None:
                path.append(curr.node_id)
                curr = curr.parent

            path.reverse()

            print("Solution path:", " - ".join(path))

            cost = best_cost[goal]

            if cost.is_integer():
                cost = int(cost)

            print("Solution cost:", cost)

            return

        # Explore neighbors
        for neighbor, edge_cost in adjlist[curr.node_id]:

            new_g = curr.g + edge_cost

            h = heuristic(neighbor, goal, coords)

            new_f = new_g + h

            new_state = NodeState(
                neighbor,
                curr,
                new_g,
                new_f
            )

            heapq.heappush(minQ, new_state)

    print("No path found")


# -----------------------------
# Read Input File
# -----------------------------

coords = {}
adjlist = {}

with open("input.txt", "r") as file:

    # Number of vertices
    V = int(file.readline())

    # Read vertices and coordinates
    for _ in range(V):

        node, x, y = file.readline().split()

        coords[node] = (float(x), float(y))

        adjlist[node] = []

    # Number of edges
    E = int(file.readline())

    # Read edges
    for _ in range(E):

        u, v, cost = file.readline().split()

        adjlist[u].append((v, float(cost)))

    # Start and goal
    start = file.readline().strip()
    goal = file.readline().strip()


# Run A* Search
a_star_search(start, goal, coords, adjlist)
```
